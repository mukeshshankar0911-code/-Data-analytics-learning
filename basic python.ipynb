```python

print("=== Animal Personality Classifier ===")
print("Answer these 3 questions to discover your animal personality!")

# Question 1
answer1 = input("\nIn a group, are you :\n1) leader\n2) independent\n3) quiet\n4) a team player? ").strip().lower()

# Question 2
answer2 = input("\nWhen solving a problem,\n1) do you take charge\n2) work alone\n3) observe first\n4) ask your team? ").strip().lower()

# Question 3
answer3 = input("\nWhich quality describes you best:\n1) brave\n2) focused\n3) observant\n4) loyal? ").strip().lower()

# Initialize scores
lion = 0
tiger = 0
panther = 0
wolf = 0

# Process answer 1
if answer1 == "leader":
    lion += 1
elif answer1 == "independent":
    tiger += 1
elif answer1 == "quiet":
    panther += 1
elif answer1 == "team player":
    wolf += 1
else:
    print("Invalid word:")

# Process answer 2
if answer2 == "take charge":
    lion += 1
elif answer2 == "work alone":
    tiger += 1
elif answer2 == "observe first":
    panther += 1
elif answer2 == "ask your team":
    wolf += 1
else:
    print("Invalid word")

# Process answer 3
if answer3 == "brave":
    lion += 1
elif answer3 == "focused":
    tiger += 1
elif answer3 == "observant":
    panther += 1
elif answer3 == "loyal":
    wolf += 1
else:
    print("Invalid word")

# Find the highest score
if lion >= tiger and lion >= panther and lion >= wolf:
    print("Your animal personality is LION!")
elif tiger >= panther and tiger >= wolf:
    print("Your animal personality is TIGER!")
elif panther >= wolf:
    print("Your animal personality is PANTHER!")
else:
    print("Your animal personality is WOLF!")
```

```python

print("=== Element Personality Classifier ===")
print("Answer these 3 questions to discover your element!")

# Question 1
answer1 = input("\nIn a difficult situation do you choose: \n1)brave \n2)calm \n3)practical or \n4)creative? ").strip().lower()

# Question 2
answer2 = input("\nIn a team are you:\n1) energetic \n2)patient \n3)reliable or \n4)imaginative? ").strip().lower()

# Question 3
answer3 = input("\nWhich quality describes you best: \n1)confident \n2)flexible \n3)organized or \n4)curious? ").strip().lower()

# Initialize scores
fire = 0
water = 0
earth = 0
air = 0

# Check the first answer
if answer1 == "brave":
    fire += 1
elif answer1 == "calm":
    water += 1
elif answer1 == "practical":
    earth += 1
elif answer1 == "creative":
    air += 1

# Check the second answer
if answer2 == "energetic":
    fire += 1
elif answer2 == "patient":
    water += 1
elif answer2 == "reliable":
    earth += 1
elif answer2 == "imaginative":
    air += 1

# Check the third answer
if answer3 == "confident":
    fire += 1
elif answer3 == "flexible":
    water += 1
elif answer3 == "organized":
    earth += 1
elif answer3 == "curious":
    air += 1

# Find the highest score
if fire >= water and fire >= earth and fire >= air:
    print("Your element personality is FIRE! 🔥")
elif water >= earth and water >= air:
    print("Your element personality is WATER! 💧")
elif earth >= air:
    print("Your element personality is EARTH! 🌿")
else:
    print("Your element personality is AIR! 🌬️")
```

```python
# 1. Dictionary: Store employee information

employees = [
    {"name": "Mukesh", "salary": 30000},
    {"name": "Arun", "salary": 25000},
    {"name": "Kumar", "salary": 40000},
    {"name": "Ravi", "salary": 20000}
]


# 2. Function: Analyse employee salaries

def salary_analysis(employees):

    total_salary = 0

    # 3. Loop: Go through each employee

    for employee in employees:

        salary = employee["salary"]

        # 4. Condition: Check salary

        if salary >= 30000:
            print(employee["name"], "- High salary")

        else:
            print(employee["name"], "- Regular salary")

        # 5. Operator: Calculate total

        total_salary += salary

    print("Total salary:", total_salary)


# 6. Call the function

salary_analysis(employees)
```

```python
def sales_analysis(sales):

    total = sum(sales)
    average = total / len(sales)
    highest = max(sales)
    lowest = min(sales)

    print("Total sales:", total)
    print("Average sales:", average)
    print("Highest sales:", highest)
    print("Lowest sales:", lowest)


monthly_sales = [10000, 15000, 20000, 25000, 30000]

sales_analysis(monthly_sales)
```
